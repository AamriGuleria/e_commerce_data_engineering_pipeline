```python
def inspect_csv(df):
    print("Shape:")
    print(df.shape)

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nFirst 5 rows:")
    print(df.head())

    print("\nData types:")
    print(df.dtypes)

    print("\nMissing values:")
    print(df.isnull().sum())

    print("\nDuplicate rows:")
    print(df.duplicated().sum())
```

```python
import pandas as pd
from pathlib import Path

if __name__ == "__main__":
    working_directory = Path.cwd().resolve()
    dataset_path = next(
        (
            directory / "dataset" / "Brazilian E-Commerce Public Dataset by Olist.csv"
            for directory in (working_directory, *working_directory.parents)
            if (directory / "dataset" / "Brazilian E-Commerce Public Dataset by Olist.csv").exists()
        ),
        None,
    )
    if dataset_path is None:
        raise FileNotFoundError("Could not locate the dataset directory from the notebook working directory")

    raw_tables = ["customers", "products", "sellers", "orders", "payments"]
    source_data = pd.read_csv(dataset_path)
    inspect_csv(source_data)
```
